```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://books.toscrape.com/"
```

```python
response = requests.get(url)

print(response.status_code)
```

```python
soup = BeautifulSoup(response.text, 'html.parser')
```

```python
books = soup.find_all('article', class_='product_pod')

print(len(books))
```

```python
titles = []
prices = []
ratings = []

for book in books:

    title = book.h3.a['title']

    price = book.find('p', class_='price_color').text

    rating = book.find('p')['class'][1]

    titles.append(title)
    prices.append(price)
    ratings.append(rating)
```

```python
data = pd.DataFrame({
    'Title': titles,
    'Price': prices,
    'Rating': ratings
})
```

```python
print(data.head())
```

```python
data.to_csv("books_dataset.csv", index=False)

print("Dataset Saved Successfully!")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://books.toscrape.com/"

response = requests.get(url)

soup = BeautifulSoup(response.text, 'html.parser')

books = soup.find_all('article', class_='product_pod')

titles = []
prices = []
ratings = []

for book in books:

    title = book.h3.a['title']

    price = book.find('p', class_='price_color').text

    rating = book.find('p')['class'][1]

    titles.append(title)
    prices.append(price)
    ratings.append(rating)

data = pd.DataFrame({
    'Title': titles,
    'Price': prices,
    'Rating': ratings
})

print(data)

data.to_csv("books_dataset.csv", index=False)

print("Dataset Saved Successfully!")
```
